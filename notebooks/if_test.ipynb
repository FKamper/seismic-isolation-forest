This notebook contains code for testing aspects of the isolation forest.

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.ensemble import IsolationForest as IF
```

```python
n = 100000
p = 1
X = np.random.uniform(low=0, high=1, size=(n, p))
Xte = np.random.uniform(low=0, high=1, size=(256, p))


mod = IF()
mod.fit(X)
scores = -mod.score_samples(Xte)
print(np.mean(-np.log2(scores)))
```

```python
def c_theory(psi):
    H = np.log(psi - 1) + 0.5772156649  # harmonic number approximation
    return 2 * H - (2 * (psi - 1) / psi)
```

```python
psi = 256


mod = IF().fit(X)
X_ref = np.random.uniform(0, 1, size=(100000, X.shape[1]))

scores_ref = -mod.score_samples(X_ref)

c_th = c_theory(psi)

h_ref = -c_th * np.log2(scores_ref)

c_emp = np.mean(h_ref)

print("c_theory:", c_th)
print("c_emp:", c_emp)
print("ratio:", c_emp / c_th)
```

```python
scores = -mod.score_samples(Xte)

h = -c_th * np.log2(scores)

normalized = h / c_emp

print(np.mean(normalized))
```
